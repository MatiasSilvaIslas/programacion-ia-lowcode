## 1. Programa base

Ejecutá la siguiente celda y observá el resultado. No modifiques todavía el código.

````python
inventario: list[dict] = [
    {"nombre": "Lámpara", "precio": 25000.0, "stock": 8},
    {"nombre": "Silla  grande", "precio": 45000.0, "stock": 3},
    {"nombre": "Mesa", "precio": 120000.0, "stock": 0},
    {"nombre": "Banco", "precio": 30000.0, "stock": 2},
]

inventario2: list[dict] = []

def mostrar_inventario(productos: list[dict]) -> None:
    if len(productos) == 0:
        print("No hay productos registrados.")
        return

    print("Productos registrados:")

    for producto in productos:
        print(
            f"- {producto['nombre']} | "
            f"Precio: ${producto['precio']:,.2f} | "
            f"Stock: {producto['stock']}"
        )

print(mostrar_inventario.__annotations__['return'])
mostrar_inventario(inventario)
mostrar_inventario(inventario2)
````

## 3. Etapa Edit — Buscar un producto

Solicitá mediante **Edit** una función llamada `buscar_producto()` que:

- Reciba la lista de productos y un nombre.
- Ignore mayúsculas, minúsculas y espacios al comienzo o al final.
- Devuelva el diccionario del producto si lo encuentra.
- Devuelva `None` si no existe.
- No modifique `mostrar_inventario()` ni la lista original.

### Procedimiento

1. Seleccioná la celda siguiente.
2. Elegí **Edit**.
3. Escribí un prompt completo.
4. Revisá la propuesta línea por línea.
5. Aceptala solamente si cumple todos los requisitos.
6. Ejecutá la celda.

````python
# Busca un producto por nombre y devuelve su diccionario o None.
def buscar_producto(
    productos: list[dict],
    nombre_buscado: str
) -> dict | None:
    # Normalizamos el nombre para ignorar mayúsculas y espacios externos.
    nombre_normalizado: str = nombre_buscado.strip().lower()

    for producto in productos:
        nombre_producto: str = str(producto.get("nombre", "")).strip().lower()

        if nombre_producto == nombre_normalizado:
            return producto

    return None
````

## 4. Etapa Edit — Calcular el valor total

Creá con Copilot una función llamada `calcular_valor_total()` que sume, para cada producto:

```text
precio × stock
```

La función debe devolver un `float` y devolver `0.0` si la lista está vacía. Elaborá vos el prompt: no copies solamente la consigna.

````python
# Calcula y devuelve el valor económico total del inventario.
def calcular_valor_total(productos: list[dict]) -> float:
    total: float = 0.0

    for producto in productos:
        total += producto["precio"] * producto["stock"]

    return total
````

## 5. Etapa Agent — Productos con poco stock y pruebas

Pedile al modo **Agent** que complete una tarea de varios pasos:

1. Crear `productos_con_poco_stock(productos, limite)`.
2. Devolver una nueva lista con los productos cuyo stock sea menor o igual al límite.
3. No modificar la lista original.
4. Crear pruebas con `assert` para las tres funciones desarrolladas.
5. No usar bibliotecas externas.

Antes de enviarlo, agregá al prompt los casos que considerás necesarios para comprobar la solución. Revisá todas las acciones propuestas por Agent.

````python
# Devuelve una nueva lista con los productos cuyo stock no supera el límite.
def productos_con_poco_stock(
    productos: list[dict],
    limite: int
) -> list[dict]:
    return [producto for producto in productos if producto["stock"] <= limite]
````

### Pruebas automáticas — completar con Agent y revisar

Agent debe reemplazar los comentarios por pruebas con `assert`. No aceptes pruebas que solo impriman valores: deben comparar el resultado real con el esperado.

````python
# Conservamos una copia para verificar que ninguna función altere el inventario.
inventario_antes: list[dict] = [producto.copy() for producto in inventario]

# buscar_producto: producto exacto, mayúsculas/minúsculas, espacios e inexistente.
assert buscar_producto(inventario, "Lámpara") == inventario[0]
assert buscar_producto(inventario, "lÁMpArA") == inventario[0]
assert buscar_producto(inventario, "  Lámpara  ") == inventario[0]
assert buscar_producto(inventario, "Estante") is None

# calcular_valor_total: varios productos, stock cero, lista vacía y tipo de retorno.
assert calcular_valor_total(inventario) == 395000.0
assert isinstance(calcular_valor_total(inventario), float)
assert calcular_valor_total(inventario2) == 0.0
assert isinstance(calcular_valor_total(inventario2), float)

# productos_con_poco_stock: por debajo, igual y por encima del límite.
productos_filtrados: list[dict] = productos_con_poco_stock(inventario, 2)
assert productos_filtrados == [inventario[2], inventario[3]]
assert inventario[2] in productos_filtrados  # Stock igual a 0, inferior al límite.
assert inventario[3] in productos_filtrados  # Stock exactamente igual al límite.
assert inventario[1] not in productos_filtrados  # Stock superior al límite.
assert productos_con_poco_stock(inventario, -1) == []
assert productos_con_poco_stock(inventario2, 2) == []
assert productos_filtrados is not inventario

# Las tres funciones deben dejar intacta la lista y sus diccionarios.
assert inventario == inventario_antes

print("Todas las pruebas assert se ejecutaron correctamente.")
````
