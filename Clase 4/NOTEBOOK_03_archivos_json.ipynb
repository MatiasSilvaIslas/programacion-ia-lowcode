# Clase 4 — Archivos JSON

JSON es un formato de texto utilizado para guardar e intercambiar datos. Permite conservar la información después de cerrar el programa. También es un formato utilizado cuando realizamos consultas a una API, lo cual veremos más adelante.

## 1. Relación entre Python y JSON

| Python | JSON |
|---|---|
| `dict` | objeto |
| `list` | arreglo |
| `str` | texto |
| `int` y `float` | número |
| `True` y `False` | `true` y `false` |
| `None` | `null` |

Python incluye el módulo `json`, por lo que no necesitamos instalar paquetes adicionales.

```python
import json
```

## 2. Crear los datos

Usaremos una lista de diccionarios para representar varios productos.

```python
productos: list[dict] = [
    {"nombre": "Lámpara", "precio": 25000.0, "stock": 8},
    {"nombre": "Silla", "precio": 45000.0, "stock": 3}
]

print(productos)
```

## 3. Guardar información

El modo `"w"` abre el archivo para escritura. Si no existe, lo crea. `indent=4` ordena el contenido y `ensure_ascii=False` conserva tildes y caracteres especiales.

```python
with open("productos_ejemplo.json", "w", encoding="utf-8") as archivo:
    json.dump(
        productos,
        archivo,
        indent=4,
        ensure_ascii=False
    )

print("Productos guardados correctamente.")
```

El archivo se crea en la carpeta desde la que se está ejecutando la notebook. Podemos consultar esa ubicación con `Path.cwd()`.

```python
from pathlib import Path

print("Carpeta de trabajo:", Path.cwd())
```

## 4. Leer la información

El modo `"r"` abre el archivo para lectura. `json.load()` transforma el contenido JSON en estructuras de Python.

```python
with open("productos_ejemplo.json", "r", encoding="utf-8") as archivo:
    productos_cargados: list[dict] = json.load(archivo)

print(productos_cargados)
```

## 5. Recorrer los datos recuperados

```python
for producto in productos_cargados:
    print(
        f"{producto['nombre']} — "
        f"${producto['precio']:,.2f} — "
        f"Stock: {producto['stock']}"
    )
```

## 6. Modificar y volver a guardar

El procedimiento habitual es: **leer → modificar → guardar**.

```python
productos_cargados.append(
    {"nombre": "Mesa", "precio": 120000.0, "stock": 2}
)

with open("productos_ejemplo.json", "w", encoding="utf-8") as archivo:
    json.dump(
        productos_cargados,
        archivo,
        indent=4,
        ensure_ascii=False
    )

print("Producto agregado y archivo actualizado.")
```

## 7. Controlar si el archivo no existe

La primera vez que ejecutamos un programa, el archivo puede no existir. En ese caso iniciamos una lista vacía.

```python
try:
    with open("archivo_que_no_existe.json", "r", encoding="utf-8") as archivo:
        datos: list[dict] = json.load(archivo)

except FileNotFoundError:
    datos = []
    print("El archivo todavía no existe. Se inicia una lista vacía.")

print(datos)
```

## 8. Separar la lectura y la escritura en funciones

Utilizamos un parámetro para decidir con qué archivo trabaja cada función.

```python
import json


# Guarda una lista de productos en un archivo JSON.
def guardar_productos(
    productos: list[dict],
    nombre_archivo: str
) -> None:
    with open(nombre_archivo, "w", encoding="utf-8") as archivo:
        json.dump(
            productos,
            archivo,
            indent=4,
            ensure_ascii=False
        )

    print(f"Productos guardados en {nombre_archivo}")


# Recupera los productos guardados.
def cargar_productos(nombre_archivo: str) -> list[dict]:
    try:
        with open(nombre_archivo, "r", encoding="utf-8") as archivo:
            productos: list[dict] = json.load(archivo)

        print(f"Productos cargados desde {nombre_archivo}")
        return productos

    except FileNotFoundError:
        print(f"El archivo {nombre_archivo} todavía no existe.")
        return []


# Nombre del archivo que utilizarán ambas funciones.
archivo_productos: str = "productos.json"


# Lista original.
productos: list[dict] = [
    {
        "nombre": "Lámpara",
        "precio": 25000.0,
        "stock": 8
    },
    {
        "nombre": "Silla",
        "precio": 45000.0,
        "stock": 3
    }
]


# 1. Guardamos la lista.
guardar_productos(productos, archivo_productos)


# 2. Recuperamos la lista desde el archivo.
productos_recuperados: list[dict] = cargar_productos(
    archivo_productos
)


# 3. Mostramos los productos recuperados.
print("\nProductos recuperados:")

for producto in productos_recuperados:
    print(
        f"- {producto['nombre']} | "
        f"Precio: ${producto['precio']:,.2f} | "
        f"Stock: {producto['stock']}"
    )
```

# Ejercicio

El programa debe:

1. Cargar los productos desde un archivo JSON.
2. Agregar un producto nuevo.
3. Guardar la lista actualizada.
4. Mostrar todos los productos.
5. Funcionar aunque el archivo todavía no exista.

```python
# Nombre del archivo donde se guardará el inventario.
# Al escribirlo una sola vez, todas las funciones utilizan el mismo archivo.
NOMBRE_ARCHIVO: str = "inventario_ejercicio.json"


# Esta función intenta recuperar el inventario desde el archivo JSON.
# Devuelve una lista de diccionarios.
def cargar_inventario() -> list[dict]:

    # El bloque try contiene una operación que podría producir un error.
    try:

        # Abrimos el archivo en modo lectura: "r".
        # UTF-8 permite leer correctamente tildes, ñ y otros caracteres.
        with open(
            NOMBRE_ARCHIVO,
            "r",
            encoding="utf-8"
        ) as archivo:

            # json.load() lee el archivo y convierte su contenido
            # en una lista de diccionarios de Python.
            inventario: list[dict] = json.load(archivo)

        # Este mensaje solo aparece si el archivo pudo leerse.
        print("Productos cargados correctamente.")

        # Entregamos la lista recuperada.
        return inventario

    # Este bloque se ejecuta si el archivo todavía no existe.
    except FileNotFoundError:

        print(
            "El archivo todavía no existe. "
            "Se inicia una lista vacía."
        )

        # Devolvemos una lista vacía para que el programa pueda continuar.
        return []


# Esta función recibe una lista de productos y la guarda en JSON.
# No devuelve un resultado, por eso utiliza -> None.
def guardar_inventario(
    inventario: list[dict]
) -> None:

    # Abrimos el archivo en modo escritura: "w".
    # Si no existe, Python lo crea.
    # Si ya existe, reemplaza su contenido.
    with open(
        NOMBRE_ARCHIVO,
        "w",
        encoding="utf-8"
    ) as archivo:

        # json.dump() transforma la lista de Python
        # y la escribe dentro del archivo JSON.
        json.dump(
            inventario,          # Datos que queremos guardar.
            archivo,             # Archivo de destino.
            indent=4,            # Indentación para facilitar la lectura.
            ensure_ascii=False   # Conserva tildes y caracteres especiales.
        )

    # Cuando termina el bloque with, el archivo se cierra automáticamente.
    print("Productos guardados correctamente.")


# Esta función recibe el inventario y muestra sus productos.
# No modifica la lista ni devuelve un resultado.
def mostrar_inventario(
    inventario: list[dict]
) -> None:

    # Comprobamos si la lista está vacía.
    if len(inventario) == 0:

        print("No hay productos registrados.")

        # return termina la función en este punto.
        # Evita ejecutar el recorrido que aparece debajo.
        return

    # Este mensaje se muestra si existe al menos un producto.
    # \n agrega un salto de línea antes del texto.
    print("\nProductos registrados:")

    # Recorremos los diccionarios guardados en la lista.
    for producto in inventario:

        # Accedemos a cada valor mediante su clave.
        print(
            f"- {producto['nombre']} | "
            f"Precio: ${producto['precio']:,.2f} | "
            f"Stock: {producto['stock']}"
        )
```

## Ejecutar la solución

Cada vez que se ejecuta esta celda se agrega un producto nuevo al archivo.

```python
inventario: list[dict] = cargar_inventario()

producto_nuevo: dict = {
    "nombre": "Lámpara",
    "precio": 25000.0,
    "stock": 8
}

inventario.append(producto_nuevo)
guardar_inventario(inventario)
mostrar_inventario(inventario)
```

## Comprobar la persistencia

Volvemos a leer el archivo para comprobar que el producto quedó guardado.

```python
inventario_recuperado: list[dict] = cargar_inventario()
mostrar_inventario(inventario_recuperado)
```
