# Clase 4 — Manejo de errores

Una **excepción** es un problema que aparece mientras Python ejecuta un programa. Con `try` y `except` podemos controlar el error y evitar que el programa se detenga abruptamente.

Esta notebook contiene ejemplos y ejercicios resueltos.

## 1. Controlar un `ValueError`

Python produce un `ValueError` cuando intentamos convertir un texto que no representa un número.

```python
dato_ingresado: str = "10"

try:
    cantidad: int = int(dato_ingresado)
    print(f"Cantidad registrada: {cantidad}")
#except ValueError as error:
#    print(error)
except ValueError:
    print("Debe especificar un número entero")
```

### Probar los dos caminos

La función permite probar un valor correcto y otro incorrecto sin utilizar `input()`.

```python
def registrar_cantidad(dato: str) -> None:
    try:
        cantidad: int = int(dato)
        print(f"Cantidad registrada: {cantidad}")
    except ValueError:
        print(f"'{dato}' no es un número entero válido.")


registrar_cantidad("12")
registrar_cantidad("doce")
```

## 2. Validar números negativos

`try/except` controla la conversión. La condición `if` comprueba si el número tiene sentido para nuestro problema.

```python
def registrar_stock(dato: str) -> None:
    try:
        stock: int = int(dato)

        if stock < 0:
            print("El stock no puede ser negativo.")
        else:
            print(f"Stock registrado: {stock}")

    except ValueError:
        print("El stock debe ser un número entero.")


registrar_stock("8")
registrar_stock("-3")
registrar_stock("1.1")
```

## 3. Evitar un `KeyError`

Un `KeyError` aparece al solicitar una clave que no existe. El método `.get()` permite indicar un valor alternativo.

```python
producto: dict = {
    "nombre": "Lámpara",
    "precio": 25000.0
}

stock: int = producto.get("stock", 0)

print(f"Producto: {producto['nombre']}")
print(f"Stock: {stock}")
```

## 4. Controlar diferentes excepciones

Cada `except` responde a un tipo de problema específico.

```python
def calcular_total_producto(producto: dict, cantidad_texto: str) -> None:
    try:
        cantidad: int = int(cantidad_texto)
        precio: float = float(producto["precio"])
        total: float = precio * cantidad

        print(f"Total: ${total:,.2f}")

    except ValueError:
        print("No fue posible convertir uno de los valores.")

    except KeyError:
        print("Falta el precio del producto.")


producto_correcto: dict = {"nombre": "Silla", "precio": 45000.0}
producto_sin_precio: dict = {"nombre": "Mesa"}

calcular_total_producto(producto_correcto, "2")
calcular_total_producto(producto_correcto, "dos")
calcular_total_producto(producto_sin_precio, "2")
```

## 5. Repetir hasta encontrar un valor válido

En un programa interactivo podemos combinar `while` con `try/except`. Para que esta notebook no quede esperando una entrada, simulamos varios intentos mediante una lista.

```python
# Simulamos los valores que un usuario podría ingresar.
intentos: list[str] = ["cinco", "-2", "5"]

# Indica si ya encontramos una cantidad correcta.
cantidad_valida: bool = False

# Indica qué elemento de la lista vamos a probar.
indice: int = 0

# Preparamos la variable donde guardaremos el resultado.
cantidad: int = 0


# El ciclo continúa mientras:
# 1. no tengamos una cantidad válida;
# 2. todavía queden intentos en la lista.
while not cantidad_valida and indice < len(intentos):

    # Tomamos el intento ubicado en la posición actual.
    dato: str = intentos[indice]

    print(f"Intento recibido: {dato}")

    # Avanzamos para que la próxima vuelta tome otro intento.
    indice += 1

    try:
        # Intentamos convertir el texto en un número entero.
        cantidad = int(dato)

        # El número existe, pero todavía debemos validarlo.
        if cantidad < 0:
            print("La cantidad no puede ser negativa.")

        else:
            # El número es entero y no es negativo.
            cantidad_valida = True

    except ValueError:
        # Se ejecuta si int(dato) no puede realizar la conversión.
        print("Debe ingresar un número entero.")


# Solo mostramos el resultado si encontramos un valor correcto.
if cantidad_valida:
    print(f"Cantidad registrada: {cantidad}")
```

# Ejercicio

Crear un programa que:

1. Reciba el precio y la cantidad de un producto.
2. Controle que los valores sean numéricos.
3. Rechace valores negativos.
4. Calcule y muestre el total.
5. Devuelva `None` cuando los datos sean incorrectos.

```python
def calcular_total(precio_texto: str, cantidad_texto: str) -> float | None:
    try:
        precio: float = float(precio_texto)
        cantidad: int = int(cantidad_texto)

        if precio < 0 or cantidad < 0:
            print("El precio y la cantidad no pueden ser negativos.")
            return None

        total: float = precio * cantidad
        print(f"Total: ${total:,.2f}")
        return total

    except ValueError:
        print("El precio y la cantidad deben ser números válidos.")
        return None
```

## Casos de prueba

Probamos un caso correcto, un texto inválido, un valor negativo y un número decimal utilizado como cantidad.

```python
print("Caso 1: datos correctos")
calcular_total("25000", "3")

print("\nCaso 2: texto no numérico")
calcular_total("veinticinco mil", "3")

print("\nCaso 3: precio negativo")
calcular_total("-25000", "3")

print("\nCaso 4: cantidad decimal")
calcular_total("25000", "3.5")
```

## Variante interactiva

> Esta celda queda esperando que el usuario escriba los valores. No está bloqueada: hay que completar los cuadros de entrada que aparecen en la parte superior de la notebook.

```python
precio_ingresado: str = input("Precio: " )
cantidad_ingresada: str = input("Cantidad: " )

calcular_total(precio_ingresado, cantidad_ingresada)
```
