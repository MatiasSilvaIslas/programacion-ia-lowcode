# 6. Formas de trabajar con GitHub Copilot

> Idea central: **pedir → revisar → ejecutar → comprobar**. Copilot propone; el programador decide qué aceptar.

# Parte A — Cómo trabajar con GitHub Copilot

## 1. Cuatro formas de usar Copilot

| Modalidad | Qué hace | Control del programador |
|---|---|---|
| Autocompletado | Sugiere código mientras escribimos | Aceptamos con `Tab` o rechazamos con `Esc` |
| Ask | Explica y analiza el código | No debería modificar archivos |
| Edit | Propone un cambio concreto | Revisamos y aceptamos las diferencias |
| Agent | Resuelve una tarea más completa | Puede modificar archivos y ejecutar comandos |

En esta primera parte veremos cómo acceder a cada modalidad, qué escribir y qué debemos revisar antes de aceptar una propuesta.

## 2. Preparar VS Code y abrir Copilot

Antes de comenzar, debemos tener instalada la extensión **GitHub Copilot** y haber iniciado sesión con una cuenta de GitHub que tenga Copilot habilitado.

### Paso a paso

1. Abrí en VS Code la carpeta completa de la clase mediante `Archivo → Abrir carpeta`.
2. Abrí esta notebook y seleccioná el kernel **Python — Clase 4**.
3. Localizá el ícono de Copilot en la barra superior o lateral de VS Code.
4. Hacé clic en **Open el chat** o abrí la paleta con `Command + Shift + P` en macOS o `Ctrl + Shift + P` en Windows.
5. Buscá `GitHub Copilot: Open Chat`.
6. Comprobá que el panel permita elegir un modo como **Ask**, **Edit** o **Agent**.

> Los nombres y la ubicación de los botones pueden cambiar ligeramente según la versión de VS Code y de la extensión.

### ¿Qué contexto puede ver Copilot?

Copilot responde mejor cuando tiene acceso al archivo abierto, al código seleccionado y a una consigna precisa. Abrir una carpeta completa también le permite comprender la relación entre varios archivos. De todos modos, no debemos asumir que interpretó todo correctamente: conviene mencionar la función o el archivo sobre el que queremos trabajar.

## 3. Autocompletado: recibir sugerencias mientras escribimos

El autocompletado propone código directamente en el editor. La sugerencia suele aparecer en color gris, delante del cursor.

### Paso a paso

1. Creá una celda de código nueva.
2. Escribí un comentario que describa una tarea sencilla. Por ejemplo:

```python
# Crear una función que calcule el precio con IVA
```

3. Presioná `Enter` y esperá unos segundos.
4. Si aparece una propuesta en gris, leela antes de incorporarla.
5. Presioná `Tab` para aceptar la sugerencia o `Esc` para rechazarla.
6. Ejecutá la celda con valores de prueba.

### Qué debemos revisar

- Que la función haga exactamente lo solicitado.
- Que los nombres de variables sean claros.
- Que no utilice contenidos que todavía no vimos.
- Que el resultado sea correcto para más de un valor.

El comentario actúa como una instrucción breve. Cuanto más claro sea, más útil suele ser la propuesta.

````python
# Crear una función que calcule el precio con IVA
def calcular_precio_con_iva(precio, iva):
    """
    Calcula el precio final con IVA incluido.

    :param precio: Precio base del producto (float)
    :param iva: Porcentaje de IVA (float)
    :return: Precio final con IVA (float)
    """
    precio_final = precio + (precio * iva / 100)
    return precio_final 

print(calcular_precio_con_iva(9999999999999, 21))  # Ejemplo de uso
````

## 4. Ask: preguntar y comprender sin modificar

El modo **Ask** se utiliza para pedir explicaciones, analizar errores o consultar alternativas. Es el punto de partida más conveniente cuando todavía no entendemos el código.

### Paso a paso

1. Seleccioná la celda o el fragmento que querés analizar.
2. Abrí Copilot Chat.
3. Formulá una pregunta específica e indicá que no querés cambios en el archivo.
4. Leé la explicación y comparala con el código real.
5. Si la respuesta es demasiado general, agregá contexto o hacé una segunda pregunta.

### Ejemplo de prompt

```text
Explicame paso a paso qué hace la función cargar_inventario().
Indicá qué recibe, qué devuelve y qué ocurre si el archivo no existe.
No modifiques el código.
```

Ask también sirve para consultar un error. En ese caso conviene incluir el mensaje completo, explicar qué esperábamos que ocurriera y señalar en qué celda apareció.

## 5. Edit: solicitar un cambio concreto

El modo **Edit** propone modificaciones sobre el código seleccionado o sobre un archivo. Es apropiado cuando entendemos el programa y sabemos qué cambio puntual necesitamos.

### Paso a paso

1. Seleccioná el código que Copilot puede modificar.
2. Describí el cambio solicitado.
3. Indicá expresamente qué partes no debe modificar.
4. Pedí que conserve el estilo, el tipado y los conceptos utilizados en clase.
5. Enviá el prompt y esperá la propuesta.
6. Revisá las diferencias: las líneas agregadas, eliminadas y reemplazadas.
7. Aceptá el cambio solo si respeta la consigna.
8. Ejecutá nuevamente el código y probá distintos casos.

### Ejemplo de prompt

```text
Modificá la función mostrar_inventario() para indicar el valor total del stock.
No cambies las otras funciones.
Conservá las anotaciones de tipo y no uses bibliotecas externas.
Explicá qué líneas agregaste.
```

Aceptar una edición no demuestra que funcione: después debemos ejecutarla y comprobar el resultado.

## 6. Agent: delegar una tarea completa

El modo **Agent** puede recorrer el proyecto, modificar varios archivos y, según la configuración, ejecutar comandos. Se utiliza para tareas que tienen varios pasos relacionados.

### Paso a paso

1. Abrí la carpeta completa del proyecto en VS Code.
2. Verificá que no haya cambios importantes sin guardar.
3. Abrí Copilot Chat y seleccioná **Agent**.
4. Definí el resultado que querés obtener.
5. Indicá qué archivos puede modificar y cuáles debe conservar.
6. Especificá restricciones: bibliotecas permitidas, versión de Python y conceptos que puede utilizar.
7. Pedí una forma concreta de validación, como casos de prueba.
8. Revisá el plan o las acciones que propone.
9. Prestá atención antes de autorizar comandos o cambios de archivos.
10. Revisá todas las modificaciones y ejecutá las pruebas.

### Ejemplo de prompt

```text
Agregá al programa una función para buscar productos por nombre.
Trabajá solamente sobre inventario.py y no cambies las funciones existentes.
Usá Python estándar, conservá el tipado y no agregues dependencias.
Creá casos de prueba para producto encontrado, inexistente y lista vacía.
Ejecutá las pruebas y resumí los cambios.
```

Agent tiene más autonomía, pero el programador sigue siendo responsable de delimitar la tarea, revisar y decidir.

## 7. Cómo construir un buen prompt

Una instrucción completa puede organizarse en cuatro partes:

1. **Contexto:** qué programa tenemos y cómo funciona actualmente.
2. **Tarea:** qué resultado concreto necesitamos.
3. **Restricciones:** qué no debe cambiar y qué herramientas puede usar.
4. **Comprobación:** cómo sabremos que la respuesta funciona.

### De un prompt impreciso a uno verificable

Prompt impreciso:

```text
Mejorá el inventario.
```

Prompt verificable:

```text
Tengo un inventario representado mediante una lista de diccionarios.
Agregá una función para buscar un producto por nombre.
La búsqueda no debe distinguir mayúsculas de minúsculas.
No modifiques las funciones existentes ni uses bibliotecas externas.
Probá producto encontrado, inexistente y lista vacía.
```

El segundo prompt reduce ambigüedades y permite comprobar objetivamente el resultado.

# Ejemplo

## 8. Programa base

Ejecutamos primero este inventario en memoria. No usa `input()` ni modifica archivos, por lo que la demostración se puede repetir sin bloquear la notebook.

````python
inventario: list[dict] = [
    {"nombre": "Lámpara", "precio": 25000.0, "stock": 8},
    {"nombre": "Silla", "precio": 45000.0, "stock": 3},
    {"nombre": "Mesa", "precio": 120000.0, "stock": 0},
]


def mostrar_inventario(productos: list[dict]) -> None:
    if len(productos) == 0:
        print("No hay productos registrados.")
        return

    print("Productos registrados:")
    valor_total_stock: float = 0.0

    for producto in productos:
        print(
            f"- {producto['nombre']} | "
            f"Precio: ${producto['precio']:,.2f} | "
            f"Stock: {producto['stock']}"
        )
        valor_total_stock += producto["precio"] * producto["stock"]

    print(f"\nValor total del stock: ${valor_total_stock:,.2f}")


def buscar_producto(
    productos: list[dict],
    nombre_buscado: str
) -> dict | None:
    # Normalizamos el texto para comparar nombres sin distinguir mayúsculas.
    nombre_normalizado: str = nombre_buscado.strip().lower()

    for producto in productos:
        nombre_producto: str = str(producto.get("nombre", "")).strip().lower()

        if nombre_producto == nombre_normalizado:
            return producto

    # Si no hay coincidencias, devolvemos None.
    return None


mostrar_inventario(inventario)
````

## 9. Demostración de Ask

Abrí Copilot Chat, y enviá este prompt:

```text
Analizá la función mostrar_inventario() de esta notebook.
Explicá qué recibe, qué devuelve, para qué sirve el return
y qué ocurre cuando la lista está vacía.
No modifiques el código.
```

### Qué revisar en la respuesta

- Debe reconocer que recibe una lista de diccionarios.
- Debe indicar que `-> None` significa que no devuelve un valor.
- Debe explicar que `return` termina la función cuando la lista está vacía.
- No debe afirmar que la función modifica el inventario.

Segundo prompt para mostrar que Ask también puede detectar limitaciones:

```text
¿Qué limitaciones tiene este programa de inventario?
Mencioná al menos tres y no escribas todavía la solución.
```

## 10. Demostración de Edit

Seleccioná la celda del programa base y usá este prompt:

```text
Agregá una función llamada buscar_producto que reciba el inventario
y un nombre. Debe buscar sin distinguir mayúsculas de minúsculas
y devolver el diccionario encontrado o None.

No modifiques mostrar_inventario() ni la lista original.
No uses bibliotecas externas.
Agregá comentarios breves y explicá los cambios.
```

Antes de aceptar, revisá que Copilot no haya cambiado código que estaba fuera de la tarea. La siguiente celda muestra una solución posible.

````python
def buscar_producto(
    productos: list[dict],
    nombre_buscado: str
) -> dict | None:
    # Normalizamos el texto para comparar sin distinguir mayúsculas.
    nombre_normalizado: str = nombre_buscado.strip().lower()

    for producto in productos:
        nombre_producto: str = str(producto.get("nombre", "")).strip().lower()

        if nombre_producto == nombre_normalizado:
            return producto

    # Si termina el recorrido sin coincidencias, no encontró el producto.
    return None
````

### Probar manualmente el cambio

No alcanza con que el código se vea correcto. Ejecutamos casos diferentes y observamos el resultado.

````python
producto_encontrado: dict | None = buscar_producto(inventario, "silla")
producto_inexistente: dict | None = buscar_producto(inventario, "Banco")
busqueda_con_espacios: dict | None = buscar_producto(inventario, "  LÁMPARA  ")

print("Encontrado:", producto_encontrado)
print("Inexistente:", producto_inexistente)
print("Con espacios y mayúsculas:", busqueda_con_espacios)
````

## 11. Demostración de Agent

Ahora pedimos una tarea más completa. En **Agent**, usamos:

```text
Revisá la función buscar_producto() y creá pruebas automáticas para estos casos:
1. producto encontrado;
2. producto inexistente;
3. lista vacía;
4. búsqueda con mayúsculas y espacios.

Usá solamente assert, sin bibliotecas externas.
No modifiques la función ni los datos del inventario.
Ejecutá las pruebas y explicá el resultado.
```

El pedido aclara la tarea, los casos obligatorios, las restricciones y la forma de comprobar el resultado.

````python
# Caso 1: producto encontrado.
resultado: dict | None = buscar_producto(inventario, "Silla")
assert resultado is not None
assert resultado["nombre"] == "Silla"

# Caso 2: producto inexistente.
assert buscar_producto(inventario, "Banco") is None

# Caso 3: lista vacía.
assert buscar_producto([], "Silla") is None

# Caso 4: mayúsculas y espacios no alteran la búsqueda.
resultado_normalizado: dict | None = buscar_producto(
    inventario,
    "  lámPARa  "
)
assert resultado_normalizado is not None
assert resultado_normalizado["nombre"] == "Lámpara"

print("Las cuatro pruebas se ejecutaron correctamente.")
````

### ¿Qué hace `assert`?

`assert` comprueba que una condición sea verdadera. Si lo es, el programa continúa. Si es falsa, Python produce un `AssertionError`. Por eso, ver el mensaje final confirma que todas las comprobaciones anteriores fueron superadas.

## 12. Ejemplo de autocompletado

Creá una celda vacía y escribí solamente este comentario:

```python
# Crear una función que calcule el valor total del inventario
```

Presioná `Enter` y esperá la sugerencia en gris. `Tab` la acepta y `Esc` la rechaza. Antes de aceptarla, verificá que multiplique `precio * stock` para cada producto y sume los resultados.

````python
# Creá una función que calcule el valor total del inventario.
def calcular_valor_total_inventario(productos: list[dict]) -> float:
    return sum(producto["precio"] * producto["stock"] for producto in productos)
valor_total: float = calcular_valor_total_inventario(inventario)
print(f"Valor total del inventario: ${valor_total:,.2f}")
````

````python
# Una solución de referencia para comparar con el autocompletado.
def calcular_valor_total(productos: list[dict]) -> float:
    total: float = 0.0

    for producto in productos:
        total += producto["precio"] * producto["stock"]
 
    return total


valor_total: float = calcular_valor_total(inventario)
print(f"Valor total del inventario: ${valor_total:,.2f}")
````
