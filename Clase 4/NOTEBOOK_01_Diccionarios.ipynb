# Clase 4 — Diccionarios

Un diccionario guarda información mediante pares de **clave → valor**. A diferencia de una lista, accedemos a cada dato mediante su nombre y no mediante una posición.

## 1. Crear un diccionario

En este ejemplo, `nombre`, `precio` y `stock` son las claves.

```python
producto: dict = {
    "nombre": "Lámpara",
    "precio": 25000.0,
    "stock": 8
}

print(producto)
```

```python
cliente: dict = {
    "nombre": "María González",
    "email": "maria@example.com",
    "edad": 28,
    "activo": True
}

print(cliente)
```

## 2. Acceder a los datos

Escribimos la clave entre corchetes para obtener su valor.

```python
print(producto["nombre"])
print(producto["precio"])
print(producto["stock"])
```

## 3. Modificar un valor

Asignamos un valor nuevo a una clave existente.

```python
producto["stock"] = 7

print(producto)
```

## 4. Agregar un dato nuevo

Si la clave todavía no existe, Python la incorpora al diccionario.

```python
producto["categoria"] = "Iluminación"
producto["ID"] = 23

print(producto)
```

## 5. Recorrer un diccionario

Un `for` puede recorrer las claves del diccionario.

```python
for clave in producto:
    print(clave)
```

Con `.items()` obtenemos al mismo tiempo la clave y el valor.

```python
print(producto.items())


for clave, valor in producto.items():
    print(f"Si la clave es {clave}, el valor es {valor}")
```

## 6. Utilizar `get()`

Si buscamos con corchetes una clave inexistente, el programa produce un `KeyError`. El método `get()` permite indicar un valor alternativo.

```python
color: str = producto.get("color", "Sin especificar")

print(color)
```

# Listas de diccionarios

Podemos utilizar una lista para reunir varios productos. Cada elemento de la lista es un diccionario.

```python
productos: list[dict] = [
    {"nombre": "Lámpara", "precio": 25000.0, "stock": 8},
    {"nombre": "Silla", "precio": 45000.0, "stock": 3},
    {"nombre": "Mesa", "precio": 120000.0, "stock": 0}
]

print(productos)
```

## 7. Recorrer los productos

```python
for producto_actual in productos:
    print(
        f"{producto_actual['nombre']} — "
        f"${producto_actual['precio']:,.2f} — "
        f"Stock: {producto_actual['stock']}"
    )
```

## 8. Buscar productos sin stock

Combinamos un recorrido `for` con una condición `if`.

```python
for producto_actual in productos:
    if producto_actual["stock"] == 0:
        print(f"Sin stock: {producto_actual['nombre']}")
```

# Ejercicio

Crear un diccionario que represente una actividad y realizar estas operaciones:

1. Mostrar el nombre de la actividad.
2. Cambiar la cantidad de participantes.
3. Agregar una clave llamada `duracion`.
4. Recorrer y mostrar todos los datos.
5. Calcular el presupuesto por participante.

```python
actividad: dict = {
    "nombre": "Taller de impresión 3D",
    "presupuesto": 150000.0,
    "participantes": 20
}

# Escribí tu solución debajo de este comentario.
print(f"Nombre de la actividad: {actividad['nombre']}")
print(f"Cantidad de participante:  {actividad['participantes']}")
actividad["participantes"] = 0
print(f"Cantidad de participante:  {actividad['participantes']}")
actividad["duracion"] = "4 semanas"
for clave, valor in actividad.items():
    print(f"{clave}: {valor}")

participantes: int = actividad.get("participantes", 0)

presupuesto_por_participante: float = (
    actividad.get("presupuesto", 0.0) / participantes
    if participantes != 0
    else 0.0
)

print(f"Presupuesto por participante: ${presupuesto_por_participante:,.2f}")
```

## Solución de referencia

Abrir esta sección después de intentar resolver el ejercicio.

```python
print(actividad["nombre"])

actividad["participantes"] = 12
actividad["duracion"] = "2 horas"

for clave, valor in actividad.items():
    print(f"{clave}: {valor}")

costo_por_persona: float = (
    actividad["presupuesto"] / actividad["participantes"]
)

print(f"Costo por participante: ${costo_por_persona:,.2f}")
```
