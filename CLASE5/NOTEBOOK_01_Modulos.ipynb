# Clase 5 — Módulos en Python

Un **módulo** es un archivo `.py` que contiene funciones, variables o clases que podemos utilizar desde otro programa. Sirve para organizar el código y reutilizarlo sin copiarlo.

En esta carpeta tenemos:


## 1. Importar un módulo de la biblioteca estándar

Python ya incluye módulos. `math`, por ejemplo, contiene funciones matemáticas.

````python
import math

# Importamos el módulo math para usar funciones matemáticas.
# math.sqrt() calcula la raíz cuadrada de un número.
raiz: float = math.sqrt(81)
print(f"Raíz cuadrada: {raiz}")

# math.pi guarda el valor de pi.
print(f"Valor de pi: {math.pi:.4f}")
````

## 2. Dos formas de importar

Con `import math` escribimos el nombre del módulo antes de la función. Con `from math import sqrt` importamos solamente un nombre.

````python
from math import ceil, sqrt

# Traemos solo las funciones que necesitamos del módulo math.
# sqrt() calcula la raíz cuadrada.
# ceil() redondea un número hacia arriba.
print(sqrt(144))
print(ceil(4.2))
````

## 3. Importar nuestro propio módulo

El archivo `inventario.py` está en la misma carpeta que esta notebook. Importamos sus funciones igual que las de una biblioteca.

````python
from inventario import (
    buscar_producto,
    calcular_valor_total,
    productos_con_poco_stock,
)

# Importamos las funciones definidas en inventario.py.
# Así reutilizamos la lógica sin volver a escribir el código.

productos: list[dict] = [
    {"nombre": "Lámpara", "precio": 25000.0, "stock": 8},
    {"nombre": "Silla", "precio": 45000.0, "stock": 3},
    {"nombre": "Mesa", "precio": 120000.0, "stock": 0},
]

# Creamos una lista de diccionarios; cada uno representa un producto con nombre, precio y stock.
# Buscamos un producto por su nombre.
print(buscar_producto(productos, "silla"))

# Calculamos el valor total del inventario usando el precio y la cantidad en stock.
print(f"Valor total: ${calcular_valor_total(productos):,.2f}")

# Mostramos los productos cuyo stock es menor o igual a 3.
print(productos_con_poco_stock(productos, 3))
````

## 4. Importar el módulo completo

También podemos importar el archivo completo. En ese caso, cada función se usa con la forma `modulo.funcion()`. Esto permite ver claramente de dónde proviene cada nombre.

````python
import inventario

# Importamos el módulo completo para usar sus funciones con el prefijo inventario.
resultado: dict | None = inventario.buscar_producto(productos, "Mesa")

# Buscamos el producto Mesa usando la función del módulo importado.
print(resultado)
````

## 5. Recargar un módulo en Jupyter

Jupyter mantiene los módulos importados en memoria. Si modificamos `inventario.py`, podemos reiniciar el kernel o recargar el módulo.

````python
import importlib

# importlib.reload() vuelve a cargar el módulo desde su archivo.
# Es útil en Jupyter cuando cambiamos inventario.py y queremos probar la nueva versión.
importlib.reload(inventario)

# Si no se recarga, Python puede seguir usando la versión vieja que quedó en memoria.
print("Módulo recargado.")
````
