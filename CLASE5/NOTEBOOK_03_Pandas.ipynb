# Clase 5 — Análisis de datos con pandas

`pandas` permite trabajar con datos organizados en tablas. Su estructura principal es el **DataFrame**, similar a una hoja de cálculo.

## 1. Importar pandas

Por convención, se utiliza el alias `pd`.

```python
# Importamos pathlib para trabajar con rutas de archivos y pandas para manipular tablas.
from pathlib import Path


# 'pd' es el alias estándar que se usa para pandas.import 
import pandas as pd
```

## 2. Leer un archivo CSV

El CSV de respaldo contiene datos meteorológicos de Buenos Aires. `parse_dates` convierte la columna de fecha al tipo adecuado.

```python
# Definimos la ruta del archivo CSV dentro de la carpeta 'datos'.
ruta_csv: Path = Path("datos") / "clima_buenos_aires.csv"

# Leemos el archivo y convertimos la columna 'fecha' a tipo fecha.
clima: pd.DataFrame = pd.read_csv(
    ruta_csv,
    parse_dates=["fecha"],
)

clima
# Mostramos la tabla completa para inspeccionar los datos.
```

## 3. Explorar la tabla

- `head()` muestra las primeras filas.
- `shape` informa filas y columnas.
- `columns` devuelve los nombres de columnas.
- `dtypes` muestra los tipos de datos.

```python
# Mostramos las primeras filas para ver la forma de los datos.
print("Primeras filas:")
display(clima.head())

# shape devuelve cantidad de filas y columnas; columns muestra los nombres de columnas.
print("Dimensiones:", clima.shape)
print("Columnas:", list(clima.columns))

print(clima.dtypes)

# dtypes informa el tipo de cada columna del DataFrame.print("Tipos:")
```

## 4. Seleccionar columnas y filas

Una columna se selecciona con corchetes. Una condición produce una serie de valores `True` y `False` que permite filtrar filas.

```python
# Seleccionamos una sola columna y mostramos sus primeros valores.
temperaturas_maximas: pd.Series = clima["temperatura_maxima"]
print(temperaturas_maximas.head())

# Filtramos las filas donde la precipitación fue mayor que cero.

# Es decir: mostramos los días que tuvieron lluvia.dias_con_lluvia
dias_con_lluvia: pd.DataFrame = clima[clima["precipitacion"] > 8]
print("Días con lluvia:")
display(dias_con_lluvia)
```

## 5. Calcular estadísticas

Podemos aplicar operaciones a una columna completa.

```python
# Calculamos varias estadísticas sobre la columna de temperaturas máximas y precipitaciones.
promedio_maximas: float = clima["temperatura_maxima"].mean()
maxima_registrada: float = clima["temperatura_maxima"].max()
precipitacion_total: float = clima["precipitacion"].sum()

# Mostramos los resultados con formato numérico.
print(f"Promedio de máximas: {promedio_maximas:.2f} °C")
print(f"Máxima registrada: {maxima_registrada:.2f} °C")
print(f"Precipitación total: {precipitacion_total:.2f} mm")
```

## 6. Crear una columna calculada

La amplitud térmica diaria es la diferencia entre las temperaturas máxima y mínima.

```python
# Creamos una nueva columna con la diferencia entre la temperatura máxima y mínima.
clima["amplitud_termica"] = (
    clima["temperatura_maxima"]
    - clima["temperatura_minima"]
)


# Mostramos solo la fecha y la amplitud térmica para verificar el cálculo.clima[["fecha", "amplitud_termica"]].head()
print("Amplitud térmica:")
display(clima[["fecha", "amplitud_termica"]].head())
```

## 7. Ordenar datos

Ordenamos de mayor a menor precipitación.

```python
# Ordenamos el DataFrame desde la mayor precipitación hasta la menor.
clima_ordenado: pd.DataFrame = clima.sort_values(
    by="precipitacion",
    ascending=False,
)


# Mostramos los primeros registros para ver los días más lluviosos.clima_ordenado.head()
print("Dias mas lluviosos:")
display(clima_ordenado.head(10))
```

# Ejercicio

1. Calculá el promedio de las temperaturas mínimas.
2. Filtrá los días con máxima menor que `13 °C`.
3. Buscá el día de mayor amplitud térmica.
4. Mostrá fecha, máxima, mínima y amplitud de ese día.

```python
# En este ejercicio podés calcular el promedio de las temperaturas mínimas,
def calcular_promedio_minimas(df: pd.DataFrame) -> float:
    return df["temperatura_minima"].mean()

print(f"Promedio de mínimas: {calcular_promedio_minimas(clima):.2f} °C")


# filtrar los días con máxima menor a 13°C
def filtrar_dias_maxima_menor_a_13(df: pd.DataFrame) -> pd.DataFrame:
    return df[df["temperatura_maxima"] < 13]        

print("Días con máxima menor a 13°C:")
display(filtrar_dias_maxima_menor_a_13(clima))  

#Día con mayor amplitud térmica
def dia_mayor_amplitud_termica(df: pd.DataFrame) -> pd.Series:
    return df.loc[df["amplitud_termica"].idxmax()]  

print("Día con mayor amplitud térmica:")
display(dia_mayor_amplitud_termica(clima))

#Mostrá fecha, máxima, mínima y amplitud de ese día.
dia = dia_mayor_amplitud_termica(clima) 
print(f"Fecha: {dia['fecha']}, Máxima: {dia['temperatura_maxima']}, Mínima: {dia['temperatura_minima']}, Amplitud: {dia['amplitud_termica']}")
```
