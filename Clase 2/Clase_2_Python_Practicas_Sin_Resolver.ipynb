## Una condición cambia qué instrucciones se ejecutan (`if`, `elif`, `else`)

```python
producto: str = "Lámpara"
precio_unitario: float = 25000.0
cantidad: int = 3
porcentaje_descuento: float = 10.0
presupuesto: float = 65000.0

subtotal: float = precio_unitario * cantidad
descuento: float = subtotal * porcentaje_descuento / 100
total: float = subtotal - descuento

if total <= presupuesto:
    estado: str = "Dentro del presupuesto"
elif total <= presupuesto * 1.10:
    estado = "Requiere un ajuste menor"
else:
    estado = "Supera el presupuesto"

print(f"Producto: {producto}")
print(f"Subtotal: ${subtotal:.2f}")
print(f"Descuento: ${descuento:.2f}")
print(f"Total final: ${total:.2f}")
print(f"Presupuesto disponible: ${presupuesto:.2f}")
print(f"Estado: {estado}")
```

# Práctica 3: Ampliar la regla

## Consigna
Agregá `horas_disponibles` y decidí si la producción es viable **por presupuesto y por tiempo**.

**Expansión:** Conservá mensajes específicos para saber cuál de los dos criterios falla.

**Comprobar:** Probar cuatro combinaciones de `True` y `False`.

```python
# Escribí tu solución acá
producto: str = "Lámpara" 
precio_unitario: float = 25000.0
cantidad: int = 3

porcentaje_descuento: float = 10.0
presupuesto: float = 65000.0

horas_necesarias: float = 8.0
horas_disponibles: float = 10.0

subtotal: float = precio_unitario * cantidad
descuento: float = subtotal * porcentaje_descuento / 100
total: float = subtotal - descuento

presupuesto_ok: bool = total <= presupuesto
tiempo_ok: bool = horas_necesarias <= horas_disponibles

if presupuesto_ok and tiempo_ok:
    resultado: str = "Producción viable"
elif not presupuesto_ok and tiempo_ok:
    resultado = "Producción no viable: supera el presupuesto"
elif presupuesto_ok and not tiempo_ok:
    resultado = "Producción no viable: no hay suficiente tiempo"
else:
    resultado = "Producción no viable: supera el presupuesto y no hay suficiente tiempo"
print(f"Producto: {producto}")
print(f"Subtotal: ${subtotal:.2f}")
print(f"Descuento: ${descuento:.2f}") 
print(f"Total final: ${total:.2f}")
print(f"Presupuesto disponible: ${presupuesto:.2f}")
print(f"Horas necesarias: {horas_necesarias}")
print(f"Horas disponibles: {horas_disponibles}")
print(f"Resultado: {resultado}")
```

# Práctica 4: Obtener información de una lista

## Consigna
Calculá el **promedio de los costos** recorriendo la lista.

**Expansión:** Encontrá el costo mayor con un recorrido propio y compará después con `max()`.

**Comprobar:** Probar también una lista vacía.

```python
# Escribí tu solución acá
costos: list[float] = [
    45000.0,
    19200.0,
    8500.0,
]

total: float = 0.0
for costo in costos: 
    total += costo

promedio: float = total / len(costos)
print(f"El promedio es: ${promedio:,.2f}")


costos2: list[float] = []

total2: float = 0.0
for costo in costos2:
    total2 += costo

if len(costos2) == 0:
    print("No puede calcularse el promedio de una lista vacía")
else:
    promedio2: float = total2 / len(costos2)
    print(f"El promedio es: ${promedio2:,.2f}")
```

## Una función le da nombre a una operación reutilizable

```python
costos: list[float] = [
    45000.0,
    19200.0,
    8500.0,
]
def calcular_total(costos: list[float]) -> float:
    total: float = 0.0
    for costo in costos:
        total += costo
    return total

total_produccion: float = calcular_total(costos)
print(f"Total de la producción: ${total_produccion:,.2f}")
```

## Separar funciones permite cambiar una regla sin reescribir todo

**Expansión del material:** escribir una tercera función que produzca el mensaje final.

```python
def aplicar_descuento(subtotal: float, porcentaje: float) -> float:
    descuento: float = subtotal * porcentaje / 100
    return subtotal - descuento

def entra_en_presupuesto(total: float, presupuesto: float) -> bool:
    return total <= presupuesto

subtotal: float = 75000.0
porcentaje_descuento: float = 10.0
presupuesto: float = 70000.0
total_final: float = aplicar_descuento(
    subtotal,
    porcentaje_descuento
)
presupuesto_ok: bool = entra_en_presupuesto(
    total_final,
    presupuesto
)
print(f"Subtotal: ${subtotal:,.2f}")
print(f"Total final: ${total_final:,.2f}")
print(f"¿Entra en el presupuesto?: {presupuesto_ok}")
```

## Clases: datos y métodos

```python
# "class" indica que vamos a crear una clase.
# "Producto" es el nombre de la clase.
class Producto:
    # __init__ se ejecuta automáticamente al crear un producto.
    def __init__(
        self,
        nombre: str,
        precio_unitario: float,
        cantidad: int
    ) -> None:
        # Guarda los datos recibidos dentro del objeto.
        self.nombre: str = nombre
        self.precio_unitario: float = precio_unitario
        self.cantidad: int = cantidad

    # Define una acción que podrá realizar cada producto.
    def calcular_subtotal(self) -> float:
        # Devuelve el precio multiplicado por la cantidad.
        return self.precio_unitario * self.cantidad
```

```python
lampara: Producto = Producto(
    nombre="Lámpara",
    precio_unitario=25000.0,
    cantidad=3
)

# Accedemos a sus atributos
print(f"Producto: {lampara.nombre}")
print(f"Precio unitario: ${lampara.precio_unitario:,.2f}")
print(f"Cantidad: {lampara.cantidad}")

# Ejecutamos uno de sus métodos
subtotal: float = lampara.calcular_subtotal()
print(f"Subtotal: ${subtotal:,.2f}")
```
