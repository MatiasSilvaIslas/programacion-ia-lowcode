## 1. Listas
Una lista reúne varios valores que necesitamos recorrer. `len()` devuelve su cantidad de elementos; `append()` agrega uno.

```python
costos: list[float] = [
    45000.0,
    19200.0,
    8500.0,
]

print(costos[0])
print(len(costos))
```

**Para probar:** la lista puede crecer con `costos.append(12000.0)`.

## 2. Tuplas
Una tupla reúne datos relacionados en posiciones fijas: índice 0, nombre del rubro; índice 1, costo.

```python
rubro: tuple[str, float] = (
    "Espacio",
    120000.0,
)

print(rubro[0])
print(rubro[1])
```

Una lista puede contener varias tuplas: `rubros: list[tuple[str, float]]`.

## 3. `for`: recorrido y acumulación
El bloque indentado se repite para cada elemento.

```python
costos: list[float] = [45000.0, 19200.0, 8500.0]

total: float = 0.0
for costo in costos:
    total += costo
    print(f"Procesado: ${costo:,.2f}")

print(f"Total: ${total:,.2f}")
```

## 4. Desempaquetado de tuplas
`for nombre, costo in rubros` permite utilizar ambos valores. `_` indica que el primer valor no se necesita.

El ejemplo de la diapositiva usa `rubros` y `total` ya definidos. Para ejecutarlo independientemente, inicializamos esos datos en una celda anterior.

```python
rubros: list[tuple[str, float]] = [("Espacio", 120000.0), ("Materiales", 45000.0)]
total: float = 0.0
```

```python
for nombre, costo in rubros:
    print(nombre)
    print(costo)

for _, costo in rubros:
    total += costo
```

## 5. `while`: repetición con condición
La condición debe poder cambiar para que el bucle termine.

```python
pendientes: int = 3
while pendientes > 0:
    print(f"Quedan {pendientes}")
    pendientes -= 1

print("Proceso terminado")
```

## 6. Funciones: parámetros y retorno
Una función encapsula una operación reutilizable.

```python
costos: list[float] = [
    45000.0,
    19200.0,
    8500.0,
]

def calcular_total(costos: list[float]) -> float:
    total: float = 0.0
    for costo in costos:
        total += costo
    return total

total_produccion: float = calcular_total(costos)
print(f"Total de la producción: ${total_produccion:,.2f}")
```

## 8. Clases: datos y métodos
`__init__` inicializa los datos del producto y `calcular_subtotal` es un método.

```python
# "class" indica que vamos a crear una clase.
class Producto:
    # __init__ se ejecuta al crear un producto.
    def __init__(
        self,
        nombre: str,
        precio_unitario: float,
        cantidad: int,
    ) -> None:
        # Guarda los datos dentro del objeto.
        self.nombre: str = nombre
        self.precio_unitario: float = precio_unitario
        self.cantidad: int = cantidad

    # Define una acción del producto.
    def calcular_subtotal(self) -> float:
        return self.precio_unitario * self.cantidad
```

```python
lampara: Producto = Producto(
    nombre="Lámpara",
    precio_unitario=25000.0,
    cantidad=3,
)

print(f"Producto: {lampara.nombre}")
print(f"Precio unitario: ${lampara.precio_unitario:,.2f}")
print(f"Cantidad: {lampara.cantidad}")

subtotal: float = lampara.calcular_subtotal()
print(f"Subtotal: ${subtotal:,.2f}")
```
