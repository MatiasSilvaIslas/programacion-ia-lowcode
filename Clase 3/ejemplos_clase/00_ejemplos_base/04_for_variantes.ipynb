# 4. Variantes de for
Cuando una lista contiene tuplas, `for` puede separar sus valores. El guion bajo `_` se usa para indicar que un valor no será utilizado.

```python
# crea la lista rubros con los valores [
rubros: list[tuple[str, float]] = [
    # ejecuta: ("Espacio", 120000.0),
    ("Espacio", 120000.0),
    # ejecuta: ("Técnica", 95000.0),
    ("Técnica", 95000.0),
    # ejecuta: ("Comunicación", 38000.0),
    ("Comunicación", 38000.0),
# ejecuta: ]
]

# Usamos los dos valores de cada tupla.
# itera sobre rubros usando nombre, costo como elemento
for nombre, costo in rubros:
    # muestra por pantalla el valor indicado
    print(f"{nombre}: ${costo:,.2f}")

# Usamos solamente el costo.
# crea la variable total con valor 0.0
total: float = 0.0
# itera sobre rubros usando _, costo como elemento
for _, costo in rubros:
    # ejecuta: total += costo
    total += costo

# muestra por pantalla el valor indicado
print(f"Total: ${total:,.2f}")
```
