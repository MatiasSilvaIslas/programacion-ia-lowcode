# 2. Tuplas
Una tupla reúne datos relacionados en posiciones fijas.

```python
# crea la variable rubro con valor ("Espacio", 120000.0)
rubro: tuple[str, float] = ("Espacio", 120000.0)

# crea la variable nombre con valor rubro[0]
nombre: str = rubro[0]
# crea la variable costo con valor rubro[1]
costo: float = rubro[1]

# muestra por pantalla el valor indicado
print(f"Rubro: {nombre}")
# muestra por pantalla el valor indicado
print(f"Costo: ${costo:,.2f}")
```
