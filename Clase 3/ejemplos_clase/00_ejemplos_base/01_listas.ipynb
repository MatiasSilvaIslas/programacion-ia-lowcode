# 1. Listas
Una lista reúne varios valores y puede crecer durante la ejecución.

```python
# crea la lista costos con los valores [45000.0, 19200.0, 8500.0]
costos: list[float] = [45000.0, 19200.0, 8500.0]

# muestra por pantalla el valor indicado
print(f"Primer costo: ${costos[0]:,.2f}")
# muestra por pantalla el valor indicado
print(f"Cantidad de costos: {len(costos)}")

# añade 12000.0 a costos
costos.append(12000.0)
# muestra por pantalla el valor indicado
print(f"Lista actualizada: {costos}")
```
