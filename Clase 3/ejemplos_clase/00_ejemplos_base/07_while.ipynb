# 7. While
`while` repite un bloque mientras una condición sea verdadera.


```python
# crea la variable pendientes con valor 3
pendientes: int = 3

# repite mientras pendientes > 0
while pendientes > 0:
    # muestra por pantalla el valor indicado
    print(f"Quedan {pendientes} tareas")
    # ejecuta: pendientes -= 1
    pendientes -= 1

# muestra por pantalla el valor indicado
print("Proceso terminado")
```
