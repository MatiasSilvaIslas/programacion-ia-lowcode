# 5. Funciones
Una función le da nombre a una operación reutilizable. Recibe datos mediante parámetros y devuelve un resultado con `return`.

```python
# ejecuta: def calcular_total(costos: list[float]) -> float:
def calcular_total(costos: list[float]) -> float:
    # crea la variable total con valor 0.0
    total: float = 0.0

    # itera sobre costos usando costo como elemento
    for costo in costos:
        # ejecuta: total += costo
        total += costo

    # devuelve total
    return total


# crea la lista costos con los valores [45000.0, 19200.0, 8500.0]
costos: list[float] = [45000.0, 19200.0, 8500.0]
# crea la variable total_produccion con valor calcular_total(costos)
total_produccion: float = calcular_total(costos)

# muestra por pantalla el valor indicado
print(f"Total de la producción: ${total_produccion:,.2f}")
```
