# 6. Clases y objetos
Una clase reúne los datos y las acciones de un tipo de objeto.

```python
# ejecuta: class Producto:
class Producto:
    # ejecuta: def __init__(
    def __init__(
        # ejecuta: self,
        self,
        # ejecuta: nombre: str,
        nombre: str,
        # ejecuta: precio_unitario: float,
        precio_unitario: float,
        # ejecuta: cantidad: int,
        cantidad: int,
    # ejecuta: ) -> None:
    ) -> None:
        # ejecuta: self.nombre: str = nombre
        self.nombre: str = nombre
        # ejecuta: self.precio_unitario: float = precio_unitario
        self.precio_unitario: float = precio_unitario
        # ejecuta: self.cantidad: int = cantidad
        self.cantidad: int = cantidad

    # ejecuta: def calcular_subtotal(self) -> float:
    def calcular_subtotal(self) -> float:
        # devuelve self.precio_unitario * self.cantidad
        return self.precio_unitario * self.cantidad


# crea la variable lampara con valor Producto(
lampara: Producto = Producto(
    # crea la variable nombre con valor "Lámpara modular",
    nombre="Lámpara modular",
    # crea la variable precio_unitario con valor 25000.0,
    precio_unitario=25000.0,
    # crea la variable cantidad con valor 3,
    cantidad=3,
# ejecuta: )
)

# crea la variable subtotal con valor lampara.calcular_subtotal()
subtotal: float = lampara.calcular_subtotal()

# muestra por pantalla el valor indicado
print(f"Producto: {lampara.nombre}")
# muestra por pantalla el valor indicado
print(f"Cantidad: {lampara.cantidad}")
# muestra por pantalla el valor indicado
print(f"Subtotal: ${subtotal:,.2f}")
```
