# 3. For simple
`for` repite una acción para cada elemento de una lista.

```python
# crea la lista costos con los valores [45000.0, 19200.0, 8500.0]
costos: list[float] = [45000.0, 19200.0, 8500.0]
# crea la variable total con valor 0.0
total: float = 0.0

# itera sobre costos usando costo como elemento
for costo in costos:
    # ejecuta: total += costo
    total += costo
    # muestra por pantalla el valor indicado
    print(f"Procesado: ${costo:,.2f}")

# muestra por pantalla el valor indicado
print(f"Total: ${total:,.2f}")
```
