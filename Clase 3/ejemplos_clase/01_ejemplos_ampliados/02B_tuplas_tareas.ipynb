# Tuplas · Planificación de tareas
Las tuplas permiten reunir el nombre de una tarea, su duración y su prioridad.


## Analizar una planificación
Calculamos el tiempo total e informamos las tareas prioritarias.


```python
# crea la lista tareas con los valores [
tareas: list[tuple[str, float, bool]] = [
    # ejecuta: ("Diseñar prototipo", 4.5, True),
    ("Diseñar prototipo", 4.5, True),
    # ejecuta: ("Comprar materiales", 2.0, False),
    ("Comprar materiales", 2.0, False),
    # ejecuta: ("Construir estructura", 8.0, True),
    ("Construir estructura", 8.0, True),
    # ejecuta: ("Documentar proceso", 3.0, False),
    ("Documentar proceso", 3.0, False),
# ejecuta: ]
]

# crea la variable horas_totales con valor 0.0
horas_totales: float = 0.0
horas_totales_priritarias = 0.0

# itera sobre tareas usando nombre, horas, prioritaria como elemento
for nombre, horas, prioritaria in tareas:
    # ejecuta: horas_totales += horas
    horas_totales += horas

    # ejecuta el bloque si prioritaria
    if prioritaria:
        # muestra por pantalla el valor indicado
        print(f"PRIORITARIA · {nombre}: {horas} horas")
        horas_totales_priritarias += horas

# muestra por pantalla el valor indicado
print(f"Horas totales del proyecto: {horas_totales}")
print(f"Horas totales prioritarias del proyecto: {horas_totales_priritarias}")
```
