# Clases · Planificador de producción
El objeto representa una producción y administra materiales, unidades y tiempo disponible.


## Modelo completo de producción
Los métodos separan costo de materiales, horas necesarias y viabilidad.


```python
# ejecuta: class Produccion:
class Produccion:
    # ejecuta: def __init__(
    def __init__(
        # ejecuta: self,
        self,
        # ejecuta: nombre: str,
        nombre: str,
        # ejecuta: unidades: int,
        unidades: int,
        # ejecuta: presupuesto: float,
        presupuesto: float,
        # ejecuta: horas_disponibles: float,
        horas_disponibles: float,
    # ejecuta: ) -> None:
    ) -> None:
        # ejecuta: self.nombre: str = nombre
        self.nombre: str = nombre
        # ejecuta: self.unidades: int = unidades
        self.unidades: int = unidades
        # ejecuta: self.presupuesto: float = presupuesto
        self.presupuesto: float = presupuesto
        # ejecuta: self.horas_disponibles: float = horas_disponibles
        self.horas_disponibles: float = horas_disponibles
        # ejecuta: self.materiales: list[tuple[str, float]] = []
        self.materiales: list[tuple[str, float]] = []

    # ejecuta: def agregar_material(self, nombre: str, costo_unitario: float) -> None:
    def agregar_material(self, nombre: str, costo_unitario: float) -> None:
        # añade (nombre, costo_unitario) a self.materiales
        self.materiales.append((nombre, costo_unitario))

    # ejecuta: def calcular_materiales(self) -> float:
    def calcular_materiales(self) -> float:
        # crea la variable costo_unitario con valor 0.0
        costo_unitario: float = 0.0

        # itera sobre self.materiales usando _, costo como elemento
        for _, costo in self.materiales:
            # ejecuta: costo_unitario += costo
            costo_unitario += costo

        # devuelve costo_unitario * self.unidades
        return costo_unitario * self.unidades

    # ejecuta: def calcular_horas(self, horas_por_unidad: float) -> float:
    def calcular_horas(self, horas_por_unidad: float) -> float:
        # devuelve horas_por_unidad * self.unidades
        return horas_por_unidad * self.unidades

    def calcular_maximo_unidades(self, horas_por_unidad: float) -> int:
        return int(self.horas_disponibles // horas_por_unidad)

    # ejecuta: def evaluar(self, horas_por_unidad: float) -> str:
    def evaluar(self, horas_por_unidad: float) -> str:
        # crea la variable presupuesto_ok con valor self.calcular_materiales() <= self.presupuesto
        presupuesto_ok: bool = self.calcular_materiales() <= self.presupuesto
        # crea la variable tiempo_ok con valor self.calcular_horas(horas_por_unidad) <= self.horas_disponibles
        tiempo_ok: bool = self.calcular_horas(horas_por_unidad) <= self.horas_disponibles

        # ejecuta el bloque si presupuesto_ok and tiempo_ok
        if presupuesto_ok and tiempo_ok:
            # devuelve "Producción viable"
            return "Producción viable"
        # ejecuta el bloque si not presupuesto_ok and not tiempo_ok
        if not presupuesto_ok and not tiempo_ok:
            # devuelve "Revisar presupuesto y tiempo"
            return "Revisar presupuesto y tiempo"
        # ejecuta el bloque si not presupuesto_ok
        if not presupuesto_ok:
            # devuelve "Revisar presupuesto"
            return "Revisar presupuesto"
        # devuelve "Revisar tiempo"
        return "Revisar tiempo"


# crea la variable produccion con valor Produccion(
produccion: Produccion = Produccion(
    # crea la variable nombre con valor "Lámpara modular",
    nombre="Lámpara modular",
    # crea la variable unidades con valor 20,
    unidades=20,
    # crea la variable presupuesto con valor 650000.0,
    presupuesto=650000.0,
    # crea la variable horas_disponibles con valor 50.0,
    horas_disponibles=50.0,
# ejecuta: )
)

# ejecuta: produccion.agregar_material("Madera", 19200.0)
produccion.agregar_material("Madera", 19200.0)
# ejecuta: produccion.agregar_material("Cable", 4200.0)
produccion.agregar_material("Cable", 4200.0)
# ejecuta: produccion.agregar_material("Portalámpara", 5600.0)
produccion.agregar_material("Portalámpara", 5600.0)

# crea la variable horas_por_unidad con valor 2.2
horas_por_unidad: float = 2.2

# muestra por pantalla el valor indicado
print(f"Producción: {produccion.nombre}")
# muestra por pantalla el valor indicado
print(f"Materiales: ${produccion.calcular_materiales():,.2f}")
# muestra por pantalla el valor indicado
print(f"Horas: {produccion.calcular_horas(horas_por_unidad)}")
# muestra por pantalla el valor indicado
print(produccion.evaluar(horas_por_unidad))
print(f"Máximo de unidades: {produccion.calcular_maximo_unidades(horas_por_unidad)}")
```
