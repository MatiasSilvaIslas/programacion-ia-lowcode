# Funciones · Presupuesto con descuento
Separamos el cálculo del subtotal, el descuento y la evaluación del presupuesto.


## Definir funciones con una responsabilidad
Cada función responde una pregunta concreta.


```python
# ejecuta: def calcular_subtotal(precio: float, cantidad: int) -> float:
def calcular_subtotal(precio: float, cantidad: int) -> float:
    # devuelve precio * cantidad
    return precio * cantidad


# ejecuta: def aplicar_descuento(subtotal: float, porcentaje: float) -> float:
def aplicar_descuento(subtotal: float, porcentaje: float) -> float:
    # crea la variable descuento con valor subtotal * porcentaje / 100
    descuento: float = subtotal * porcentaje / 100
    # devuelve subtotal - descuento
    return subtotal - descuento


# ejecuta: def evaluar_presupuesto(total: float, presupuesto: float) -> str:
def evaluar_presupuesto(total: float, presupuesto: float) -> str:
    # ejecuta el bloque si total <= presupuesto
    if total <= presupuesto:
        # crea la variable diferencia con valor presupuesto - total
        diferencia: float = presupuesto - total
        # devuelve f"Viable: quedan ${diferencia:,.2f}"
        return f"Viable: quedan ${diferencia:,.2f}"

    # crea la variable diferencia con valor total - presupuesto
    diferencia = total - presupuesto
    # devuelve f"No viable: faltan ${diferencia:,.2f}"
    return f"No viable: faltan ${diferencia:,.2f}"


def costo_unidad_descuento (total: float, cantidad: int) -> float:
    return total / cantidad

# crea la variable precio con valor 25000.0
precio: float = 25000.0
# crea la variable cantidad con valor 3
cantidad: int = 3
# crea la variable porcentaje_descuento con valor 10.0
porcentaje_descuento: float = 10.0
# crea la variable presupuesto con valor 70000.0
presupuesto: float = 70000.0

# crea la variable subtotal con valor calcular_subtotal(precio, cantidad)
subtotal: float = calcular_subtotal(precio, cantidad)
# crea la variable total con valor aplicar_descuento(subtotal, porcentaje_descuento)
total: float = aplicar_descuento(subtotal, porcentaje_descuento)
# crea la variable estado con valor evaluar_presupuesto(total, presupuesto)
estado: str = evaluar_presupuesto(total, presupuesto)
costo_unidad_con_descuento: float = costo_unidad_descuento(total,cantidad)

# muestra por pantalla el valor indicado
print(f"Subtotal: ${subtotal:,.2f}")
# muestra por pantalla el valor indicado
print(f"Total: ${total:,.2f}")
# muestra por pantalla el valor indicado
print(estado)
print(f"Precio unitario con descuento: ${costo_unidad_con_descuento:,.2f}")
```
