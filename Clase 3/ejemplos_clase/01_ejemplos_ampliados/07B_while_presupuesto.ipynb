# While · Control de presupuesto
El programa incorpora costos mientras exista dinero disponible y queden elementos por revisar.


```python
# crea la lista costos con los valores [25000.0, 18000.0, 32000.0, 15000.0]
costos: list[float] = [25000.0, 18000.0, 32000.0, 15000.0]
# crea la variable presupuesto con valor 60000.0
presupuesto: float = 60000.0
# crea la variable acumulado con valor 0.0
acumulado: float = 0.0
# crea la variable indice con valor 0
indice: int = 0

# repite mientras indice < len(costos) and acumulado + costos[indice] <= presupuesto
while indice < len(costos) and acumulado + costos[indice] <= presupuesto:
    # crea la variable costo_actual con valor costos[indice]
    costo_actual: float = costos[indice]
    # ejecuta: acumulado += costo_actual
    acumulado += costo_actual
    # muestra por pantalla el valor indicado
    print(f"Costo aceptado: $ {costo_actual:,.2f}")
    # ejecuta: indice += 1
    indice += 1

# muestra por pantalla el valor indicado
print(f"Total utilizado: $ {acumulado:,.2f}")
# muestra por pantalla el valor indicado
print(f"Costos aceptados: {indice}")
# muestra por pantalla el valor indicado
print(f"Costos pendientes: {len(costos) - indice}")

faltante: float = costos[indice] - (presupuesto - acumulado) 
print(f"El primer costo que no pudo incorporarse fue: $ {costos[indice]:,.2f}, faltan: $ {faltante:,.2f} ")

```
