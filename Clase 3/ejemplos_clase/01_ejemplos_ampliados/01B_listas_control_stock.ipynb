# Listas · Control de stock
Este ejemplo utiliza dos listas relacionadas para detectar productos que necesitan reposición.


## Comparar productos y cantidades
Cada posición de `productos` corresponde a la misma posición de `stock`.


```python
# crea la lista productos con los valores ["Lámpara", "Mesa", "Panel", "Silla"]
productos: list[str] = ["Lámpara", "Mesa", "Panel", "Silla"]
# crea la lista stock con los valores [8, 2, 0, 5]
stock: list[int] = [8, 2, 0, 5]
# crea la lista precios con los valores [100, 2000, 400, 800]
precios: list[float] = [100.0, 2000.0, 400.0, 800.0]
# crea la variable stock_minimo con valor 3
stock_minimo: int = 3

# itera sobre range(len(productos)) usando indice como elemento
for indice in range(len(productos)):
    # crea la variable nombre con valor productos[indice]
    nombre: str = productos[indice]
    # crea la variable cantidad con valor stock[indice]
    cantidad: int = stock[indice]
    precio: float = precios[indice]

    # ejecuta el bloque si cantidad == 0
    if cantidad == 0:
        # crea la variable estado con valor "Sin stock"
        estado: str = "Sin stock"
    # si la condición anterior fue falsa, ejecuta este bloque cuando cantidad < stock_minimo
    elif cantidad < stock_minimo:
        # crea la variable estado con valor "Reponer"
        estado = "Reponer"
    # ejecuta este bloque si ninguna condición previa se cumplió
    else:
        # crea la variable estado con valor "Stock suficiente"
        estado = "Stock suficiente"

    # muestra por pantalla el valor indicado
    print(f"{nombre}: ${precio:,.2f}, {cantidad} unidades · {estado}")
```
