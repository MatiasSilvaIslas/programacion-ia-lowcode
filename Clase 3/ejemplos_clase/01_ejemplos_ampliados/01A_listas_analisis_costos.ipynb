# Listas · Análisis de costos
Este ejemplo analiza una lista de costos y obtiene el total, el promedio y el valor más alto.


## Recorrer y analizar la lista
Acumulamos los costos y buscamos el mayor mediante un recorrido.


```python
# crea la lista costos con los valores [45000.0, 19200.0, 8500.0, 32000.0, 12700.0]
costos: list[float] = [45000.0, 19200.0, 8500.0, 32000.0, 12700.0]

#Agregar
#Calculá también cuál es el costo más bajo sin utilizar `min()`.
costo_menor: float = costos[0]

# crea la variable total con valor 0.0
total: float = 0.0
# crea la variable costo_mayor con valor costos[0]
costo_mayor: float = costos[0]

# itera sobre costos usando costo como elemento
for costo in costos:
    # ejecuta: total += costo
    total += costo

    # ejecuta el bloque si costo > costo_mayor
    if costo > costo_mayor:
        # crea la variable costo_mayor con valor costo
        costo_mayor = costo
     # ejecuta el bloque si costo < costo_mayor
    if costo < costo_menor:
        costo_menor = costo

# crea la variable promedio con valor total / len(costos)
promedio: float = total / len(costos)

#Agregar
#Calculá también cuál es el costo más bajo sin utilizar `min()`.

# muestra por pantalla el valor indicado
print(f"Cantidad de costos: {len(costos)}")
# muestra por pantalla el valor indicado
print(f"Total: ${total:,.2f}")
# muestra por pantalla el valor indicado
print(f"Promedio: ${promedio:,.2f}")
# muestra por pantalla el valor indicado
print(f"Costo mayor: ${costo_mayor:,.2f}")
# muestra por pantalla el valor indicado
print(f"Costo menor: ${costo_menor:,.2f}")


```
