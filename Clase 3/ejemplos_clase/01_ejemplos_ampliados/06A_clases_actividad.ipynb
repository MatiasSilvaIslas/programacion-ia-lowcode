# Clases · Presupuesto de una actividad
La clase guarda los datos de una actividad y sus métodos realizan los cálculos.


## Clase, objeto y métodos
Cada objeto tiene su propio presupuesto y lista de rubros.


```python
# ejecuta: class Actividad:
class Actividad:
    # ejecuta: def __init__(self, nombre: str, presupuesto: float) -> None:
    def __init__(self, nombre: str, presupuesto: float) -> None:
        # ejecuta: self.nombre: str = nombre
        self.nombre: str = nombre
        # ejecuta: self.presupuesto: float = presupuesto
        self.presupuesto: float = presupuesto
        # ejecuta: self.rubros: list[tuple[str, float]] = []
        self.rubros: list[tuple[str, float]] = []

    # ejecuta: def agregar_rubro(self, nombre: str, costo: float) -> None:
    def agregar_rubro(self, nombre: str, costo: float) -> None:
        # añade (nombre, costo) a self.rubros
        self.rubros.append((nombre, costo))

    # ejecuta: def calcular_total(self) -> float:
    def calcular_total(self) -> float:
        # crea la variable total con valor 0.0
        total: float = 0.0

        # itera sobre self.rubros usando _, costo como elemento
        for _, costo in self.rubros:
            # ejecuta: total += costo
            total += costo

        # devuelve total
        return total

    # ejecuta: def evaluar(self) -> str:
    def evaluar(self) -> str:
        # crea la variable total con valor self.calcular_total()
        total: float = self.calcular_total()

        # ejecuta el bloque si total <= self.presupuesto
        if total <= self.presupuesto:
            # devuelve "Actividad viable"
            return "Actividad viable"

        # devuelve "Actividad fuera del presupuesto"
        return "Actividad fuera del presupuesto"


    def mas_costoso(self) -> tuple[str, float]:
        rubro_maximo: tuple[str, float] = self.rubros[0]

        for rubro in self.rubros:
            if rubro[1] > rubro_maximo[1]:
                rubro_maximo = rubro

        return rubro_maximo


# crea la variable actividad con valor Actividad("Muestra de proyectos", 350000.0)
actividad: Actividad = Actividad("Muestra de proyectos", 350000.0)
# ejecuta: actividad.agregar_rubro("Espacio", 120000.0)
actividad.agregar_rubro("Espacio", 120000.0)
# ejecuta: actividad.agregar_rubro("Técnica", 95000.0)
actividad.agregar_rubro("Técnica", 95000.0)
# ejecuta: actividad.agregar_rubro("Comunicación", 38000.0)
actividad.agregar_rubro("Comunicación", 38000.0)
# ejecuta: actividad.agregar_rubro("Catering", 72000.0)
actividad.agregar_rubro("Catering", 72000.0)

# muestra por pantalla el valor indicado
print(f"Actividad: {actividad.nombre}")
# muestra por pantalla el valor indicado
print(f"Total: ${actividad.calcular_total():,.2f}")
# muestra por pantalla el valor indicado
print(f"Presupuesto: ${actividad.presupuesto:,.2f}")
# muestra por pantalla el valor indicado
print(actividad.evaluar())

rubro = actividad.mas_costoso()

print(f"Rubro más costoso: {rubro[0]}")
print(f"Costo: ${rubro[1]:,.2f}")
```
