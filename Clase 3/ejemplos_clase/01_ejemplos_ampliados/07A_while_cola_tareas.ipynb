# While · Cola de tareas
El loop continúa mientras existan tareas pendientes en la lista.


```python
# crea la lista pendientes con los valores [
pendientes: list[str] = [
    # ejecuta: "Preparar materiales",
    "Preparar materiales",
    # ejecuta: "Construir prototipo",
    "Construir prototipo",
    # ejecuta: "Probar funcionamiento",
    "Probar funcionamiento",
    # ejecuta: "Documentar resultado",
    "Documentar resultado",
# ejecuta: ]
]
# crea la lista completadas con los valores []
completadas: list[str] = []

# repite mientras pendientes
while pendientes and len(completadas) < 3:
    # crea la variable tarea_actual con valor pendientes.pop(0)
    tarea_actual: str = pendientes.pop(0)
    # muestra por pantalla el valor indicado
    print(f"Procesando: {tarea_actual}")
    # añade tarea_actual a completadas
    completadas.append(tarea_actual)

# muestra por pantalla el valor indicado
print(f"Tareas completadas: {len(completadas)}")
# muestra por pantalla el valor indicado
print(completadas)
print(f"Pendientes: {pendientes}")
```
