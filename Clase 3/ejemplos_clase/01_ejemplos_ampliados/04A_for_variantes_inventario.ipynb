# For con tuplas · Inventario
Usamos `for nombre, stock, minimo` para separar los datos de cada producto.


## Desempaquetar tres valores
Cada vuelta compara el stock actual con el mínimo recomendado.


```python
# crea la lista inventario con los valores [
inventario: list[tuple[str, int, int]] = [
    # ejecuta: ("Lámpara", 8, 3),
    ("Lámpara", 8, 3),
    # ejecuta: ("Mesa", 2, 4),
    ("Mesa", 2, 4),
    # ejecuta: ("Panel", 0, 5),
    ("Panel", 0, 5),
    # ejecuta: ("Silla", 7, 4),
    ("Silla", 7, 4),
# ejecuta: ]
]

# crea la variable productos_a_reponer con valor 0
productos_a_reponer: int = 0
total_reponer: int = 0

# itera sobre inventario usando nombre, stock, minimo como elemento
for nombre, stock, minimo in inventario:
    # ejecuta el bloque si stock < minimo
    if stock < minimo:
        # crea la variable faltante con valor minimo - stock
        faltante: int = minimo - stock
        # ejecuta: productos_a_reponer += 1
        productos_a_reponer += 1
        # muestra por pantalla el valor indicado
        print(f"{nombre}: reponer {faltante} unidades")
        total_reponer += faltante
    # ejecuta este bloque si ninguna condición previa se cumplió
    else:
        # muestra por pantalla el valor indicado
        print(f"{nombre}: stock suficiente")

# muestra por pantalla el valor indicado
print(f"Productos que requieren reposición: {productos_a_reponer}")
print(f"Unidades a reponer: {total_reponer}")
```
