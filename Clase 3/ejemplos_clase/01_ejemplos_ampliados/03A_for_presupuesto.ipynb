# For · Presupuesto acumulado
El programa procesa los costos uno por uno y avisa cuándo se supera el presupuesto.


## Acumular y comparar
En cada vuelta actualizamos el total acumulado.


```python
# crea la lista costos con los valores [45000.0, 19200.0, 8500.0, 32000.0]
costos: list[float] = [45000.0, 19200.0, 8500.0, 32000.0]
# crea la variable presupuesto con valor 90000.0
presupuesto: float = 90000.0
# crea la variable acumulado con valor 0.0
acumulado: float = 0.0
cantidad_costos: int = 0

# itera sobre costos usando costo como elemento
for costo in costos:
    # ejecuta: acumulado += costo
    acumulado += costo
    # muestra por pantalla el valor indicado
    print(f"Costo procesado: ${costo:,.2f}")
    # muestra por pantalla el valor indicado
    print(f"Acumulado: ${acumulado:,.2f}")

    # ejecuta el bloque si acumulado > presupuesto
    if acumulado > presupuesto:
        # crea la variable diferencia con valor acumulado - presupuesto
        diferencia: float = acumulado - presupuesto
        # muestra por pantalla el valor indicado
        print(f"Presupuesto superado por ${diferencia:,.2f}")
    # ejecuta este bloque si ninguna condición previa se cumplió
    else:
        # crea la variable disponible con valor presupuesto - acumulado
        disponible: float = presupuesto - acumulado
        # muestra por pantalla el valor indicado
        print(f"Disponible: ${disponible:,.2f}")
        cantidad_costos +=1
        print(f"Cantidad de costos procesados: {cantidad_costos}")

    # muestra por pantalla el valor indicado
    print("-" * 40)
```
