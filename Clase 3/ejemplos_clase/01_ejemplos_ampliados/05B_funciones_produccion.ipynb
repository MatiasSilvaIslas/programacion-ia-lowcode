# Funciones · Viabilidad de una producción
El programa combina varias funciones para evaluar costo y tiempo.


## Combinar resultados
Una función puede utilizar el resultado de otra.


```python
# ejecuta: def calcular_costo_materiales(
def calcular_costo_materiales(
    # ejecuta: costos_unitarios: list[float],
    costos_unitarios: list[float],
    # ejecuta: unidades: int,
    unidades: int,
# ejecuta: ) -> float:
) -> float:
    # crea la variable total_unitario con valor 0.0
    total_unitario: float = 0.0

    # itera sobre costos_unitarios usando costo como elemento
    for costo in costos_unitarios:
        # ejecuta: total_unitario += costo
        total_unitario += costo

    # devuelve total_unitario * unidades
    return total_unitario * unidades


# ejecuta: def calcular_horas(unidades: int, horas_por_unidad: float) -> float:
def calcular_horas(unidades: int, horas_por_unidad: float) -> float:
    # devuelve unidades * horas_por_unidad
    return unidades * horas_por_unidad

def calcular_mano_obra(horas: float, costo_hora: float) -> float:
    return horas * costo_hora
    
# ejecuta: def evaluar_produccion(
def evaluar_produccion(
    # ejecuta: costo: float,
    costo: float,
    # ejecuta: presupuesto: float,
    presupuesto: float,
    # ejecuta: horas: float,
    horas: float,
    # ejecuta: horas_disponibles: float,
    horas_disponibles: float,
# ejecuta: ) -> str:
) -> str:
    # crea la variable presupuesto_ok con valor costo <= presupuesto
    presupuesto_ok: bool = costo <= presupuesto
    # crea la variable tiempo_ok con valor horas <= horas_disponibles
    tiempo_ok: bool = horas <= horas_disponibles

    # ejecuta el bloque si presupuesto_ok and tiempo_ok
    if presupuesto_ok and tiempo_ok:
        # devuelve "Producción viable"
        return "Producción viable"
    # ejecuta el bloque si not presupuesto_ok and not tiempo_ok
    if not presupuesto_ok and not tiempo_ok:
        # devuelve "Revisar presupuesto y tiempo"
        return "Revisar presupuesto y tiempo"
    # ejecuta el bloque si not presupuesto_ok
    if not presupuesto_ok:
        # devuelve "Revisar presupuesto"
        return "Revisar presupuesto"
    # devuelve "Revisar tiempo disponible"
    return "Revisar tiempo disponible"


# crea la lista costos_unitarios con los valores [19200.0, 4200.0, 5600.0]
costos_unitarios: list[float] = [19200.0, 4200.0, 5600.0]
# crea la variable unidades con valor 20
unidades: int = 20
# crea la variable presupuesto con valor 650000.0
presupuesto: float = 650000.0
# crea la variable horas_disponibles con valor 50.0
horas_disponibles: float = 50.0
# crea la variable horas_por_unidad con valor 2.2
horas_por_unidad: float = 2.2

costo_hora: float = 5000.0

# crea la variable costo con valor calcular_costo_materiales(costos_unitarios, unidades)
costo: float = calcular_costo_materiales(costos_unitarios, unidades)
# crea la variable horas con valor calcular_horas(unidades, horas_por_unidad)
horas: float = calcular_horas(unidades, horas_por_unidad)

#unifico los costos
costo_mano_obra = calcular_mano_obra(horas, costo_hora)

costo_total = costo + costo_mano_obra
# crea la variable resultado con valor evaluar_produccion(
resultado: str = evaluar_produccion(
    # ejecuta: costo,
    costo_total, # paso el costo total
    # ejecuta: presupuesto,
    presupuesto,
    # ejecuta: horas,
    horas,
    # ejecuta: horas_disponibles,
    horas_disponibles,
# ejecuta: )
)

# muestra por pantalla el valor indicado
print(f"Costo: ${costo:,.2f}")
# muestra por pantalla el valor indicado
print(f"Horas necesarias: {horas}")
# muestra por pantalla el valor indicado
print(f"Resultado: {resultado}")
```
