# Tuplas · Costo de materiales
Cada tupla guarda el nombre, la cantidad y el precio de un material.


## Lista de tuplas
El recorrido separa los tres valores de cada material.


```python
# crea la lista materiales con los valores [
materiales: list[tuple[str, float, float]] = [
    # ejecuta: ("Madera", 3.5, 24000.0),
    ("Madera", 3.5, 24000.0),
    # ejecuta: ("Cable", 8.0, 2800.0),
    ("Cable", 8.0, 2800.0),
    # ejecuta: ("Portalámpara", 4.0, 5600.0),
    ("Portalámpara", 4.0, 5600.0),
    ("Quiricocho", 3.0, 1600.0),
# ejecuta: ]
]

# crea la variable total con valor 0.0
total: float = 0.0

# guardo el maximo
producto_maximo: tuple[str, float, float] = materiales[0]
subtotal_maximo: float = 0.0

# itera sobre materiales
for material in materiales:
    nombre, cantidad, precio = material
    # crea la variable subtotal con valor cantidad * precio
    subtotal: float = cantidad * precio
    if subtotal > subtotal_maximo:
        subtotal_maximo = subtotal
        producto_maximo = material
    # ejecuta: total += subtotal
    total += subtotal
    # muestra por pantalla el valor indicado
    print(f"{nombre}: {cantidad} × ${precio:,.2f} = ${subtotal:,.2f}")

# muestra por pantalla el valor indicado
print(f"Costo total: ${total:,.2f}")
print(f"Material con mayor subtotal: {producto_maximo[0]}")
print(f"Subtotal máximo: ${subtotal_maximo:,.2f}")
```
