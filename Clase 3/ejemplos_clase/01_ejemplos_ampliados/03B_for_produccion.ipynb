# For · Producción semanal
Este ejemplo recorre una lista de unidades producidas y compara cada día con una meta.


## Evaluar resultados diarios
El loop calcula el total y cuenta los días que alcanzaron la meta.


```python
# crea la lista produccion_diaria con los valores [12, 18, 15, 9, 20]
produccion_diaria: list[int] = [12, 18, 15, 9, 20]
# crea la variable meta_diaria con valor 15
meta_diaria: int = 15
# crea la variable total con valor 0
total: int = 0
# crea la variable dias_cumplidos con valor 0
dias_cumplidos: int = 0

# itera sobre produccion_diaria usando unidades como elemento
for unidades in produccion_diaria:
    # ejecuta: total += unidades
    total += unidades

    # ejecuta el bloque si unidades >= meta_diaria
    if unidades >= meta_diaria:
        # ejecuta: dias_cumplidos += 1
        dias_cumplidos += 1
        # crea la variable estado con valor "Meta alcanzada"
        estado: str = "Meta alcanzada"
    # ejecuta este bloque si ninguna condición previa se cumplió
    else:
        # crea la variable estado con valor "Debajo de la meta"
        estado = "Debajo de la meta"

    # muestra por pantalla el valor indicado
    print(f"Producción: {unidades} unidades · {estado}")

# crea la variable promedio con valor total / len(produccion_diaria)
promedio: float = total / len(produccion_diaria)

# muestra por pantalla el valor indicado
print("-" * 45)
# muestra por pantalla el valor indicado
print(f"Producción total: {total} unidades")
# muestra por pantalla el valor indicado
print(f"Promedio diario: {promedio:.1f} unidades")
# muestra por pantalla el valor indicado
print(f"Días con meta alcanzada: {dias_cumplidos}")

dias: list[str] = ["lunes", "martes","miercoles","jueves","viernes"]
produccion_maxima: int = produccion_diaria[0]
dia_max: int = 0

for indice, unidades in enumerate(produccion_diaria):
    if unidades > produccion_maxima:
        produccion_maxima = unidades
        dia_max = indice

print(f"Mayor producción diaria: {produccion_maxima} unidades")
print(f"Día de mayor producción: {dias[dia_max]}")
```
