# For con guion bajo · Calcular ingresos
Cada venta guarda número, producto, cantidad y precio. Para calcular ingresos no necesitamos el número ni el nombre.


## Ignorar valores que no se utilizan
Los dos guiones bajos representan datos que existen, pero no intervienen en el cálculo.


```python
# crea la lista ventas con los valores [
ventas: list[tuple[int, str, int, float]] = [
    # ejecuta: (1, "Lámpara", 2, 48000.0),
    (1, "Lámpara", 2, 48000.0),
    # ejecuta: (2, "Mesa", 1, 92000.0),
    (2, "Mesa", 1, 92000.0),
    # ejecuta: (3, "Panel", 4, 36500.0),
    (3, "Panel", 4, 36500.0),
# ejecuta: ]
]

# crea la variable ingresos con valor 0.0
ingresos: float = 0.0
# crea la variable unidades_vendidas con valor 0
unidades_vendidas: int = 0

# Segundo recorrido: mostrar el detalle de las ventas
print("=" * 65)
print(f"{'REPORTE DE VENTAS':^65}")
print("=" * 65)

print(f"{'N°':<5}{'Producto':<15}{'Cantidad':>10}{'Precio':>15}{'Subtotal':>20}")
print("-" * 65)

for numero, producto, cantidad, precio in ventas:
    subtotal: float = cantidad * precio

    print(
        f"{numero:<5}"
        f"{producto:<15}"
        f"{cantidad:>10}"
        f"${precio:>14,.2f}"
        f"${subtotal:>19,.2f}"
    )

# Resumen general
print("-" * 65)



# itera sobre ventas usando _, _, cantidad, precio como elemento
for _, _, cantidad, precio in ventas:
    # ejecuta: ingresos += cantidad * precio
    ingresos += cantidad * precio
    # ejecuta: unidades_vendidas += cantidad
    unidades_vendidas += cantidad

# muestra por pantalla el valor indicado
print(f"Unidades vendidas: {unidades_vendidas}")
# muestra por pantalla el valor indicado
print(f"Ingresos totales: ${ingresos:,.2f}")


```
