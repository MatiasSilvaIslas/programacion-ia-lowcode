# Trabajo con IA sobre programas de Python

En esta actividad vamos a utilizar inteligencia artificial como **asistente** para comprender, revisar y modificar programas que ya funcionan.

> La IA no reemplaza el programa completo: primero lo ejecutamos y lo comprendemos; después pedimos una ayuda concreta y comprobamos el resultado.


## Los tres programas

| Programa | Nivel de trabajo | Modificación recomendada |
|---|---|---|
| Presupuesto de una actividad | Inicial | Agregar un porcentaje de imprevistos |
| Planificador de producción | Intermedio | Incorporar precio de venta y margen |
| Gestor de pedidos y stock | Avanzado | Agregar descuento por cantidad |

Comenzaremos todos con **Presupuesto de una actividad**. Los otros programas permiten avanzar según el nivel y el tiempo disponible.


## Recorrido de trabajo

1. **Ejecutar:** correr todas las celdas y observar el resultado.
2. **Localizar:** encontrar listas, loops, funciones, clases y condiciones.
3. **Modificar:** intentar un cambio sencillo sin IA.
4. **Consultar:** pedir a la IA una explicación o modificación acotada.
5. **Comprobar:** ejecutar distintos casos de prueba.
6. **Documentar:** registrar qué se aceptó, qué se cambió y por qué.
